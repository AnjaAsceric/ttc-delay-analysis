## Cleaning ttc_delays

```python
import pandas as pd
```

```python
import numpy as np
```

```python
df_delays = pd.read_excel("ttc_delays.xlsx") #Reading the file
```

```python
df_delays.head(10) #Showing the first 10 rows
df_delays.info() #Overview of the dataset
df_delays.describe() #Statistics for numerical columns
```

```python
#Rename columns:
df_delays = df_delays.rename(columns={
    "Min Delay": "Min_Delay",
    "Min Gap": "Min_Gap"
})

#Look for missing values:
df_delays.isnull().sum()

# Fill empty cells (9520) in Bound with 'Unknown'
df_delays["Bound"] = df_delays["Bound"].fillna("Unknown")

#Drop 44 rows with missing 'Line' value
df_delays = df_delays.dropna(subset=["Line"])

#Remove duplicates
df_delays = df_delays.drop_duplicates()

#Extract month, day and day of the week from date

df_delays["Month"] = df_delays["Date"].dt.month
df_delays["Day"] = df_delays["Date"].dt.day
df_delays["Weekday"] = df_delays["Date"].dt.day_name()
```

```python
df_delays.head(10) #Showing the first 10 rows
```

```python
df_delays.to_csv("cleaned_ttc_delays.csv", index=False)
```

## Cleaning weather

```python
df_weather = pd.read_csv("weather.csv") #Reading the file
```

```python
df_weather.head(10)
```

```python
#Rename columns (remove slashes and spaces):
df_weather.columns = df_weather.columns.str.strip().str.replace(" ", "_")
df_weather.columns = (df_weather.columns.str.strip().str.replace(r"[ /]+", "_", regex=True))
df_weather.columns = (
    df_weather.columns
    .str.strip()
    .str.replace(r"[ /()]+", "_", regex=True)
)
#Look for missing values:
df_weather.isnull().sum()

#Remove duplicates
df_weather = df_weather.drop_duplicates()

#Assign the day of the week after converting Date_Time to dt
df_weather["Date_Time"] = pd.to_datetime(df_weather["Date_Time"], errors="coerce")
df_weather["Weekday"] = df_weather["Date_Time"].dt.day_name()
```

```python
#Look for missing values:
df_weather.isnull().sum()
```

```python
df_weather.info() #Overview of the dataset
df_weather.describe() #Statistics for numerical columns
```

```python
df_weather.to_csv("cleaned_weather.csv", index=False)
```

```python
## Cleaning routes
```

```python
df_routes = pd.read_csv("routes.txt")

df_routes = df_routes.drop_duplicates()

df_routes.columns = (
    df_routes.columns
    .str.strip()
    .str.replace(r"[ \-()/]+", "_", regex=True)
    .str.replace(r"_+", "_", regex=True)
)

missing_values = df_routes.isnull().sum()
print(missing_values)
```

```python
df_routes.to_excel("cleaned_routes.xlsx", index=False)
```

## Creating one Excel file with Routes, Weather and Delays sheets

```python
with pd.ExcelWriter("ttc_project_data.xlsx", engine="openpyxl") as writer:
    df_delays.to_excel(writer, sheet_name="Delays", index=False)
    df_weather.to_excel(writer, sheet_name="Weather", index=False)
    df_routes.to_excel(writer, sheet_name="Routes", index=False)
```
